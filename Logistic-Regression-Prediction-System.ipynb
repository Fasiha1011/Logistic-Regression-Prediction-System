```python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

import gradio as gr
from flask import Flask, request, render_template_string

print("All libraries imported successfully!")
```

```python
X_diabetes = df_diabetes.drop("Outcome", axis=1)
y_diabetes = df_diabetes["Outcome"]

print(X_diabetes.shape)
print(y_diabetes.shape)
```

```python
X_bank = df_bank[["age", "balance", "duration", "campaign"]]
y_bank = df_bank["y"].map({"no": 0, "yes": 1})

print(X_bank.shape)
print(y_bank.shape)
```

```python
X_train_d, X_test_d, y_train_d, y_test_d = train_test_split(
    X_diabetes, y_diabetes,
    test_size=0.2,
    random_state=42
)

print("Training data:", X_train_d.shape)
print("Testing data:", X_test_d.shape)
```

```python
X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
    X_bank, y_bank,
    test_size=0.2,
    random_state=42
)

print("Training data:", X_train_b.shape)
print("Testing data:", X_test_b.shape)
```

```python
model_diabetes = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000)
)

model_diabetes.fit(X_train_d, y_train_d)

print("Diabetes model trained successfully!")
```

```python

# Bank Marketing dataset
X_bank = df_bank[["age", "balance", "duration", "campaign"]]
y_bank = df_bank["y"].map({"no": 0, "yes": 1})

X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
    X_bank,
    y_bank,
    test_size=0.2,
    random_state=42,
    stratify=y_bank
)

model_bank = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000, class_weight="balanced")
)

model_bank.fit(X_train_b, y_train_b)

print("Weighted Bank model trained successfully!")
```

```python

# Test different bank customer inputs
test_data = pd.DataFrame([
    [25, 100, 30, 1],
    [45, 5000, 600, 1],
    [60, -200, 10, 5],
    [35, 1000, 180, 1],
    [30, 2000, 300, 2]
], columns=["age", "balance", "duration", "campaign"])

probabilities = model_bank.predict_proba(test_data)
predictions = model_bank.predict(test_data)

for i in range(len(test_data)):
    print("Customer:", test_data.iloc[i].to_dict())
    print("Prediction:", predictions[i])
    print("Probability of subscription:", round(probabilities[i][1] * 100, 2), "%")
    print("-" * 40)
```

```python
y_pred_d = model_diabetes.predict(X_test_d)

print("Diabetes Model Accuracy:", accuracy_score(y_test_d, y_pred_d) * 100, "%")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_d, y_pred_d))

print("\nClassification Report:")
print(classification_report(y_test_d, y_pred_d))
```

```python
y_pred_b = model_bank.predict(X_test_b)

print("Bank Model Accuracy:", accuracy_score(y_test_b, y_pred_b) * 100, "%")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_b, y_pred_b))

print("\nClassification Report:")
print(classification_report(y_test_b, y_pred_b))
```

```python

def predict_diabetes(Pregnancies, Glucose, BloodPressure, SkinThickness,
                     Insulin, BMI, DiabetesPedigreeFunction, Age):

    values = [[Pregnancies, Glucose, BloodPressure, SkinThickness,
               Insulin, BMI, DiabetesPedigreeFunction, Age]]

    prediction = model_diabetes.predict(values)[0]

    if prediction == 1:
        return "Prediction: Diabetes"
    else:
        return "Prediction: No Diabetes"


def predict_bank(age, balance, duration, campaign):

    values = [[age, balance, duration, campaign]]

    prediction = model_bank.predict(values)[0]

    if prediction == 1:
        return "Prediction: Customer likely to subscribe"
    else:
        return "Prediction: Customer unlikely to subscribe"


with gr.Blocks() as demo:

    gr.Markdown("# Logistic Regression Prediction System")

    with gr.Tab("Diabetes Prediction"):

        pregnancies = gr.Number(label="Pregnancies", value=1)
        glucose = gr.Number(label="Glucose", value=120)
        bp = gr.Number(label="Blood Pressure", value=70)
        skin = gr.Number(label="Skin Thickness", value=20)
        insulin = gr.Number(label="Insulin", value=80)
        bmi = gr.Number(label="BMI", value=25)
        dpf = gr.Number(label="Diabetes Pedigree Function", value=0.5)
        age = gr.Number(label="Age", value=30)

        btn1 = gr.Button("Predict Diabetes")
        output1 = gr.Textbox(label="Result")

        btn1.click(
            predict_diabetes,
            inputs=[pregnancies, glucose, bp, skin, insulin, bmi, dpf, age],
            outputs=output1
        )

    with gr.Tab("Bank Marketing Prediction"):

        bank_age = gr.Number(label="Customer Age", value=35)
        balance = gr.Number(label="Account Balance", value=1000)
        duration = gr.Number(label="Last Contact Duration (seconds)", value=180)
        campaign = gr.Number(label="Campaign Contacts", value=1)

        btn2 = gr.Button("Predict Subscription")
        output2 = gr.Textbox(label="Result")

        btn2.click(
            predict_bank,
            inputs=[bank_age, balance, duration, campaign],
            outputs=output2
        )

demo.launch()
```

```python

from flask import Flask, request, render_template_string
import pandas as pd

app = Flask(__name__)

HTML = """
<!DOCTYPE html>
<html>
<head>
    <title>Logistic Regression Prediction System</title>
    <style>
        body {
            font-family: Arial, sans-serif;
            max-width: 700px;
            margin: 30px auto;
            padding: 20px;
            background: #f4f6f8;
        }
        h1 { text-align: center; }
        .box {
            background: white;
            padding: 20px;
            margin: 20px 0;
            border-radius: 10px;
        }
        input {
            padding: 8px;
            margin: 6px 0 14px;
            width: 95%;
            box-sizing: border-box;
        }
        button {
            padding: 10px 16px;
            cursor: pointer;
        }
        .result {
            font-weight: bold;
            color: darkgreen;
            overflow-wrap: anywhere;
        }
    </style>
</head>
<body>
    <h1>Logistic Regression Prediction System</h1>

    <div class="box">
        <h2>Diabetes Prediction</h2>
        <form method="post">
            <input type="hidden" name="model" value="diabetes">

            Pregnancies:
            <input type="number" name="pregnancies" value="1" min="0" required>

            Glucose:
            <input type="number" name="glucose" value="120" min="0" required>

            Blood Pressure:
            <input type="number" name="bp" value="70" min="0" required>

            Skin Thickness:
            <input type="number" name="skin" value="20" min="0" required>

            Insulin:
            <input type="number" name="insulin" value="80" min="0" required>

            BMI:
            <input type="number" step="any" name="bmi" value="25" min="0" required>

            Diabetes Pedigree Function:
            <input type="number" step="any" name="dpf" value="0.5" min="0" required>

            Age:
            <input type="number" name="age" value="30" min="1" required>

            <button type="submit">Predict Diabetes</button>
        </form>
    </div>

    <div class="box">
        <h2>Bank Marketing Prediction</h2>
        <form method="post">
            <input type="hidden" name="model" value="bank">

            Customer Age:
            <input type="number" name="bank_age" value="35" min="18" required>

            Account Balance:
            <input type="number" name="balance" value="1000" required>

            Contact Duration (seconds):
            <input type="number" name="duration" value="180" min="0" required>

            Campaign Contacts:
            <input type="number" name="campaign" value="1" min="1" required>

            <button type="submit">Predict Subscription</button>
        </form>
    </div>

    {% if result %}
        <div class="box result">
            {{ result }}
        </div>
    {% endif %}
</body>
</html>
"""


@app.route("/", methods=["GET", "POST"])
def home():
    result = ""

    if request.method == "POST":
        try:
            selected_model = request.form.get("model")

            if selected_model == "diabetes":
                values = [[
                    float(request.form["pregnancies"]),
                    float(request.form["glucose"]),
                    float(request.form["bp"]),
                    float(request.form["skin"]),
                    float(request.form["insulin"]),
                    float(request.form["bmi"]),
                    float(request.form["dpf"]),
                    float(request.form["age"])
                ]]

                prediction = int(model_diabetes.predict(values)[0])

                result = (
                    "Prediction: Diabetes detected by the model. "
                    "This is not a medical diagnosis."
                    if prediction == 1 else
                    "Prediction: No diabetes predicted by the model. "
                    "This is not a medical diagnosis."
                )

            elif selected_model == "bank":
                # Use named columns in the same order as model training
                values = pd.DataFrame([{
                    "age": float(request.form["bank_age"]),
                    "balance": float(request.form["balance"]),
                    "duration": float(request.form["duration"]),
                    "campaign": float(request.form["campaign"])
                }])

                print("Bank input received:", values.to_dict(orient="records"))

                # Predict using the weighted bank model
                prediction = int(model_bank.predict(values)[0])
                probabilities = model_bank.predict_proba(values)[0]
                subscription_probability = probabilities[1] * 100

                result = (
                    "Prediction: Customer likely to subscribe."
                    if prediction == 1 else
                    "Prediction: Customer unlikely to subscribe."
                )

                result += (
                    f" Subscription probability: "
                    f"{subscription_probability:.2f}%"
                )

            else:
                result = "Error: Invalid prediction model selected."

        except Exception as e:
            print("Prediction error:", repr(e))
            result = f"Prediction error: {e}"

    return render_template_string(HTML, result=result)


print("Starting Logistic Regression Flask application...")
app.run(
    host="127.0.0.1",
    port=5001,
    debug=False,
    use_reloader=False,
    threaded=True
)

```
